```python
import sys
from pathlib import Path

import pandas as pd
import sklearn

cwd = Path.cwd()
project_dir = cwd.parent if cwd.name == "notebooks" else cwd
data_dir = project_dir / "data" / "raw"

print("Python:", sys.executable)
print("pandas:", pd.__version__)
print("scikit-learn:", sklearn.__version__)
print("Data dir:", data_dir)
print("Data dir present:", data_dir.is_dir())
```

```python
for path in sorted(data_dir.iterdir()):
    print("folder" if path.is_dir() else "plik", path.name)
```

```python
device_dir = data_dir / "Danmini_Doorbell"

for path in sorted(device_dir.rglob("*")):
    if path.is_file():
        print(path.relative_to(device_dir))
```

```python
sample = pd.read_csv(device_dir / "benign_traffic.csv", nrows=5)

print("Wiersze i kolumny:", sample.shape)
display(sample.iloc[:, :8])
```

```python
rows = []

for path in sorted(device_dir.rglob("*.csv")):
    columns = pd.read_csv(path, nrows=0).columns

    # Do policzenia rekordów wystarczy odczytać jedną kolumnę.
    n_records = sum(
        len(chunk)
        for chunk in pd.read_csv(path, usecols=[0], chunksize=100_000)
    )

    rows.append({
        "file": str(path.relative_to(device_dir)),
        "records": n_records,
        "columns": len(columns),
        "match": columns.equals(sample.columns),
    })

audit = pd.DataFrame(rows)
display(audit)
```
